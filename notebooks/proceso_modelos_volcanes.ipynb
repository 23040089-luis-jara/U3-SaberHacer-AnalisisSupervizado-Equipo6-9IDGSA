# Proceso de modelos supervisados: volcanes

Este notebook ejecuta el código fuente `modelo_volcanes.py`. Incluye limpieza, clasificación, regresión, optimización, evaluación y generación de gráficas.

## 1. Instalación de dependencias

Ejecuta en la terminal: `pip install -r requirements.txt`.

## 2. Carga y preparación

El script lee la hoja `Holocene Volcano List` del Excel original. La fila 2 del archivo contiene los encabezados reales. Se genera `resultados/volcanes_limpios.csv`; los valores faltantes de predictores se imputan dentro del pipeline para evitar fuga de información.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
if not (ROOT / 'modelo_volcanes.py').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
import modelo_volcanes as proyecto

df = proyecto.load_and_clean()
print('Dimensiones del dataset limpio:', df.shape)
display(df.head())
print('\nValores faltantes por columna:')
display(df.isna().sum().to_frame('Faltantes'))
```

## 3. Clasificación

Variable objetivo: `Activity Evidence`. Se comparan Regresión logística, Random Forest y Random Forest optimizado. Métricas: Accuracy, Precision macro, Recall macro y F1 macro.

```python
metricas_clasificacion = proyecto.classification(df)
display(metricas_clasificacion)
```

## 4. Regresión

Variable objetivo: `Elevation (m)`. Se comparan Regresión lineal, Random Forest y Random Forest optimizado. Métricas: MAE, RMSE y R².

```python
metricas_regresion = proyecto.regression(df)
display(metricas_regresion)
```

## 5. Gráficas y archivos generados

Las imágenes y tablas se guardan en la carpeta `resultados/`.

```python
from IPython.display import display, Image
for nombre in ['comparacion_clasificacion.png', 'matriz_confusion.png', 'errores_regresion.png', 'r2_regresion.png']:
    ruta = ROOT / 'resultados' / nombre
    if ruta.exists():
        print(nombre)
        display(Image(filename=str(ruta)))
```

## 6. Conclusión

Compara las métricas obtenidas. En clasificación, un Accuracy y F1 macro mayores suelen indicar mejor desempeño. En regresión, MAE y RMSE menores y R² mayor indican mejor ajuste. La selección final debe basarse en las métricas reales generadas por esta ejecución.
